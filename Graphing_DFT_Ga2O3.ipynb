```python
# ==========================================
# ACTUAL QUANTUM CHEMICAL DATA
# ==========================================
x_compositions = np.array([0.0, 0.0625, 0.125, 0.25]) 
x_percentages = x_compositions * 100
temperatures_C = [470, 750, 850] # from experimental figure

# Boltzmann constant in eV/K 
kB = 8.617333262e-5 

# --- Enthalpy Inputs from DFT (eV/f.u.) ---
H_beta = np.array([0, 0.1634, 0.2837, 0.5119])
H_gamma = np.array([0.2222, 0.3797, 0.49, 0.6506])

# --- Entropy Inputs (multiplied by kB) ---
S_anion = np.array([0, 0.4172, 0.7014, 1.1303]) * kB # computed just for oxygen sites, 3 per unit cell
S_cat_beta = np.array([0, 0.1431, 0.1308, 0]) * kB # computed just for octahedral sites, 1 per unit cell 
S_cat_gamma = np.array([0.7849, 1.2938, 1.5971, 1.9731]) * kB # computed including Mg, Ga, and cation vacancies

# ==========================================
# COMBINE ENTROPIES
# ==========================================
S_total_beta = S_cat_beta + S_anion
S_total_gamma = S_cat_gamma + S_anion
```

```python
# ==========================================
# SOLVER FOR CROSSOVER CONCENTRATION
# ==========================================
def find_crossover_percentage(T_K):
    """
    Finds the Mg% where G_beta == G_gamma at a given temperature in Kelvin
    using linear interpolation between the discrete data points.
    """
    # Define a fine grid of percentages for interpolation
    fine_pct = np.linspace(0.0, 25.0, 2500)
    
    # Interpolate H and S across the percentage range
    H_b_interp = interp1d(x_percentages, H_beta, kind='linear')(fine_pct)
    H_g_interp = interp1d(x_percentages, H_gamma, kind='linear')(fine_pct)
    
    S_b_interp = interp1d(x_percentages, S_total_beta, kind='linear')(fine_pct)
    S_g_interp = interp1d(x_percentages, S_total_gamma, kind='linear')(fine_pct)
    
    # Calculate continuous free energies
    G_beta_fine = H_b_interp - T_K * S_b_interp
    # Delta G relative to beta at x=0
    G_gamma_fine = H_g_interp - T_K * S_g_interp 
    
    # Find the index where the difference changes sign or is minimized
    diff = G_beta_fine - G_gamma_fine
    idx = np.argmin(np.abs(diff))
    
    # Return NaN if curves don't actually intersect in this 0-25% range
    if idx == 0 or idx == len(fine_pct) - 1:
        if np.abs(diff[idx]) > 0.01: 
            return np.nan
            
    return fine_pct[idx]

# Generate continuous temperature range (in Kelvin)
T_range_K = np.linspace(300, 1150, 200)
T_range_C = T_range_K - 273.15

# Solve for crossover concentration at each temperature
crossover_pcts = [find_crossover_percentage(T) for T in T_range_K]

# Filter out temperatures where no crossover happens within 0-25%
valid = ~np.isnan(crossover_pcts)
plot_T_C = T_range_C[valid]
plot_pct = np.array(crossover_pcts)[valid]
```

```python
# Total width is 6.8 to 7.0 inches for typical double-column journal figures
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(7.0, 3.2), dpi=300)

# --- LEFT PLOT: FREE ENERGY CURVES ---
temp_colors = ['#2b8cbe', '#f16913', '#a50f15']

beta_handles, beta_labels = [], []
gamma_handles, gamma_labels = [], []

for i, T_C in enumerate(temperatures_C):
    T_K = T_C + 273.15
    G_beta = H_beta - T_K * S_total_beta
    G_gamma = H_gamma - T_K * S_total_gamma
    
    line_beta, = ax1.plot(x_percentages, G_beta, 
                          color=temp_colors[i], 
                          linestyle='-', 
                          marker='o', 
                          markersize=4,
                          linewidth=1.2)
    beta_handles.append(line_beta)
    beta_labels.append(f'$\\beta$ ({T_C} °C)')
    
    line_gamma, = ax1.plot(x_percentages, G_gamma, 
                            color=temp_colors[i], 
                            linestyle='--', 
                            marker='s', 
                            markersize=3.5,
                            linewidth=1.2)
    gamma_handles.append(line_gamma)
    gamma_labels.append(f'$\\gamma$ ({T_C} °C)')

grid_handles = beta_handles + gamma_handles
grid_labels = beta_labels + gamma_labels

ax1.set_xlabel('Mg concentration (%)', fontsize=8, labelpad=4)
ax1.set_ylabel(r'Relative Gibbs Free Energy ($\Delta G$, eV / f.u.)', fontsize=8, labelpad=4)
ax1.set_xticks(np.arange(0, 26, 5))
ax1.tick_params(axis='both', labelsize=7)
ax1.grid(True, linestyle=':', alpha=0.4, color='#999999')
ax1.margins(y=0.18)

ax1.legend(grid_handles, grid_labels,
           ncol=2, 
           loc="upper center", 
           frameon=True, 
           framealpha=0.95,
           edgecolor='#cccccc', 
           fontsize=6.5,
           columnspacing=1.0,
           labelspacing=0.4,
           handlelength=1.8)

# --- RIGHT PLOT: PHASE DIAGRAM ---
ax2.plot(plot_pct, plot_T_C, color='#a50f15', linewidth=1.5, linestyle='-')
ax2.fill_betweenx(plot_T_C, 0, plot_pct, color='#2b8cbe', alpha=0.15)
ax2.fill_betweenx(plot_T_C, plot_pct, 25, color='#f16913', alpha=0.15)

ax2.set_xlabel('Mg concentration (%)', fontsize=8, labelpad=4)
ax2.set_ylabel('Temperature (°C)', fontsize=8, labelpad=4)
ax2.set_xlim(0, 25)
ax2.set_ylim(485, 875)
ax2.set_xticks(np.arange(0, 26, 5))
ax2.tick_params(axis='both', labelsize=7)
ax2.grid(True, linestyle=':', alpha=0.4, color='#999999')

# Annotations as specified by the user
ax2.text(8, 600, r'$\beta$-phase', fontsize=8, color='#084594', fontweight='bold', ha='center')
ax2.text(20.2, 830, r'$\gamma$-phase', fontsize=8, color='#a50f15', fontweight='bold')

plt.tight_layout()
plt.savefig('combined_journal_figure.png', bbox_inches='tight', dpi=300)
```
