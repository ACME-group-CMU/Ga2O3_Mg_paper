```python
# import dependencies to be used in this notebook
from ase.io import read, write
from ase.build import find_optimal_cell_shape, make_supercell
import numpy as np
```

All primitive cell structure files can be found in the unsubstituted structures.zip file on the GitHub repository

```python
# UPDATE THESE PATHS FOR YOUR SETUP

file_loc = r"path/to/primitivecell.cif" #change file path
output_loc_oriented = r"path/to/oriented/supercell.cif" #change file path to store oriented supercell
### see cell 4 for approximately cubic supercell file path as below ### 
#output_loc = r"path/to/supercell"+str(i)+"_sc.cif"
```

```python
structure = read(file_loc)
```

```python
#verify primitive cell vector matrix andcell parameters
structure.cell.array 
structure.cell.cellpar() 
```

```python
#Using in-built ASE function to find transformation matrix resulting in supercell closest to a cube (here i = number of repeating units in supercell)
for i in range(2,20,2):
    transformation_matrix = find_optimal_cell_shape(structure.cell, i, 'sc') 
    supercell = make_supercell(structure,transformation_matrix)
    output_loc = r"path/to/supercell"+str(i)+"_sc.cif"
    write(output_loc, supercell)
```

Matrices for $\beta$, $\alpha$, and $\kappa$ were generated using the above in-built ASE function.

Transformation Matrices used in our work for all polymorphs are as follows:

\begin{gather*}
A_{\beta}
=
\begin{pmatrix}
2 & 0 & 0\\
-4 & 4 & 0\\
0 & 0 & 2\\
\end{pmatrix}
\end{gather*}

\begin{align*}
A_{\alpha}
=
\begin{pmatrix}
2 & 0 & 0\\
-1 & 3 & -1\\
-1 & -1 & 3\\
\end{pmatrix} 
\end{align*}

\begin{align*}
A_{\kappa}
=
\begin{pmatrix}
2 & 0 & 0\\
0 & 1 & 0\\
0 & 0 & 1\\
\end{pmatrix}
\end{align*}

\begin{align*}
A_{\gamma}
=
\begin{pmatrix}
0 & 2 & 0\\
-3 & 0 & 1\\
2 & -1 & 0\\
\end{pmatrix} 
\end{align*}

Applying the aforementioned lattice transformations to the unit cells results in supercells containing 160 atoms for the $\beta$, $\alpha$, and $\gamma$ polymorphs, and 80 atoms for the $\kappa$ polymorph. These supercells serve as the basis for modeling Mg incorporation into the Ga$_2$O$_3$ lattice by substituting Ga atoms with Mg.


For the $\gamma$ polymorph, we adopt the method of constructing an oriented bulk structure as described by  Sun and Ceder$^a$, to ensure that the out-of-plane lattice vector aligns with experimentally observed orientations.

$^a$Sun, W.; Ceder, G. Efficient creation and convergence of surface slabs. *Surface Science*
__2013__, *617*, 53–59.

```python
#Define specific transformation matrix based on oriented bulk structure transformation
transf_mat = np.array([[0,2,0],[-3,0,1],[2,-1,0]]) 
```

```python
nsc = make_supercell(structure,transf_mat)
```

```python
nsc.cell.cellpar() #verify transformed cell parameters
```

```python
write(output_loc_oriented, nsc)
```
